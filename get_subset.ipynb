```python
import pandas as pd
import numpy as np
import re
import pyspark
```

```python
df = pd.read_csv(
    "data/wiki_movie_plots_deduped.csv",
    sep=",",
)

df.head()
```

```python
missing_percent = df.isnull().mean() * 100
print("\nPercentage of missing data per column:")
print(missing_percent)
```

```python
df['Director'] = df['Director'].fillna("Unknown")
df['Cast'] = df['Cast'].fillna("Unknown")
df['Genre'] = df['Genre'].fillna("Unknown")
```

```python
df.head()
```

```python
df['plot_length'] = df['Plot'].str.len()
print(df['plot_length'].describe())
```

```python
df_subset = df.sample(frac=0.3, random_state=42)  # 30% of data
df_subset = df_subset.reset_index(drop=True)
print(f'Subset size: {df_subset.shape}')
df_subset.to_csv("data/movie_subset.csv", index=False)

missing_percent = df_subset.isnull().mean() * 100
print("\nPercentage of missing data per column:")
print(missing_percent)
```
